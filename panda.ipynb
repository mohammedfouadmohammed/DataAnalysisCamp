```python
df_merge1 = pd.read_csv("F:/DataAnalystCamp/LOTR.csv")
df_merge2 = pd.read_csv("F:/DataAnalystCamp/LOTR 2.csv")
print(df_merge1,"\n------------------\n",df_merge2)
df_merge1.merge(df_merge2,how='inner', on=['FellowshipID','FirstName'])
df_merge1.merge(df_merge2,how='outer')
df_merge1.merge(df_merge2,how='left')
df_merge1.merge(df_merge2,how='right')
df_merge1.merge(df_merge2,how='cross')
```
