```python
#sets
nameOFCites1={"cairo","london","tokyo","madrid"}
print(nameOFCites1)
nameOFCites2={"liverbool","london","seol","ryaid"}
print(nameOFCites2)
print(nameOFCites1 | nameOFCites2)
print(nameOFCites1 & nameOFCites2)
print(nameOFCites1 - nameOFCites2)
print(nameOFCites2 - nameOFCites1)
print(nameOFCites1 ^ nameOFCites2)
```

```python
#bmi = (weight in pounds * 703) / (height in inches * height in inches)
weight = input('enter your weight in pounds')
height = input('enter your height in inches')

def bmi(w,h):
    bmiCal = (w * 703) / (h * h)
    print(bmiCal)
    if bmiCal <= 18.5:
        print('classification is underweight and health risk is minimal')
    elif bmiCal <= 24.9:
        print('classification is normalweight and health risk is minimal')    
    elif bmiCal <= 29.9:
        print('classification is overweight and health risk is increased')
    elif bmiCal <= 34.9:
        print('classification is obese and health risk is high')
    elif bmiCal <= 39.9:
        print('classification is severley obese and health risk is very high')
    elif bmiCal > 39.9:
        print('classification is morbidly and health risk is extremely high')
    else:
        print('enter valid inbut')
bmi(float(weight),float(height))
```

```python
###
import os
import shutil

path = "f:/New folder (4)/"

filesName = os.listdir(path)

print(filesName)


folderNames = ['prince','toons','chelsea']

for i in range(0,3):
    if not (os.path.exists(path +folderNames[i])):
        print(path +folderNames[i])
        os.makedirs(path +folderNames[i])

for n in filesName:
    if 'prince' in n and not os.path.exists(path + 'prince/' + n):
        shutil.move(path + n, path+'prince/'+n)
    elif 'toons' in n and not os.path.exists(path + 'toons/' + n):
        shutil.move(path + n, path+'toons/'+n)
    elif 'chelsea' in n and not os.path.exists(path + 'chelsea/' + n):
        shutil.move(path + n, path+'chelsea/'+n)   
```
